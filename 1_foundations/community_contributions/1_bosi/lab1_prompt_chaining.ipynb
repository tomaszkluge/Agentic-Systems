# Lab 1 Exercise: Prompt Chaining

Three LLM calls, each feeding its output into the next:
1. Pick a business area worth exploring for an Agentic AI opportunity
2. Describe a pain point in that area
3. Propose an Agentic AI solution to that pain point

```python
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(override=True)
openai = OpenAI()
```

```python
# Step 1: pick a business area

messages = [{"role": "user", "content": "Pick one business area that might be worth exploring for an Agentic AI opportunity. Respond only with the name of the business area and one sentence on why."}]

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
business_area = response.choices[0].message.content

display(Markdown(business_area))
```

```python
# Step 2: describe a pain point in that business area

messages = [{"role": "user", "content": f"""
Here is a business area:
{business_area}

Describe one specific, challenging pain point in this industry that might be ripe for an Agentic AI solution.
Explain who experiences it, why it is hard today, and what it costs the business. Keep it to one or two short paragraphs.
"""}]

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
pain_point = response.choices[0].message.content

display(Markdown(pain_point))
```

```python
# Step 3: propose an Agentic AI solution

messages = [{"role": "user", "content": f"""
Business area:
{business_area}

Pain point:
{pain_point}

Propose an Agentic AI solution to this pain point. Include:
- A short summary of the solution
- The agents involved and what each one does
- The tools, data sources or systems the agents would use
- Where a human stays in the loop
- How success would be measured
"""}]

response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
solution = response.choices[0].message.content

display(Markdown(solution))
```
