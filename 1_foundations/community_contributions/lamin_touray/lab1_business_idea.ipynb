# Lab1_business_area - Agentic AI

#### My 3 LLM calls (business area -> paint point -> agentic solution) are as follows:

```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
# Initialize the OpenAI client
load_dotenv(override=True)
openai = OpenAI()
MODEL = "gpt-4.1-mini" 

def ask_gpt(prompt):
    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
# Business area
business_area = ask_gpt(
    "What are some innovative business ideas in the healthcare operations sector that focus on improving patient care and streamlining administrative processes in hospitals and clinics?"
     "Reply with just the business area name and one sentence explaining why.")
display(Markdown(f"**Business Area:** {business_area}"))
```

```python
# Pain points
pain_points = ask_gpt(
    f"Identify the key pain points in healthcare operations that a business idea in the area of '{business_area}' would address, such as long wait times, inefficient scheduling, or lack of patient engagement."
    "Reply with a list of 3-5 pain points, each in one sentence.")
display(Markdown(f"**Pain Points:** {pain_points}"))
```

```python
# Agentic solution
agentic_solution = ask_gpt(
    f"Describe how the business idea in the area of '{business_area}' will leverage agentic solutions "
    f"to address the pain points: {pain_points}. Reply with a detailed explanation of the agentic solution."
)
display(Markdown(f"**Agentic Solution:** {agentic_solution}"))
```
